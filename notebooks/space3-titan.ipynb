## 1. Your weight on Titan
Titan's surface gravity is about 1.35 m/s² — Earth's is 9.81.

```
my_mass_kg = 60          # change to yours

earth_weight = my_mass_kg * 9.81
titan_weight = my_mass_kg * 1.35
print(f"Earth:  {earth_weight:6.0f} newtons")
print(f"Titan:  {titan_weight:6.0f} newtons — about "
      f"{titan_weight / earth_weight:.0%} of home")
```

You would weigh about a seventh of your Earth weight — and the air is one and a half times as thick. Humans on Titan could strap on wings and fly with their arms. That is not fiction; it is these two numbers.

## 2. How cold is −179 °C, really

```
titan_c = -179
titan_k = titan_c + 273.15
water_freezes_k = 273.15
methane_liquid_k = (90.7, 111.7)   # liquid range at 1 atm, kelvin

print("Titan surface:", titan_k, "K")
print("Water freezes:", water_freezes_k, "K — Titan is", 
      round(water_freezes_k - titan_k), "K below: water is ROCK there")
print("Methane is liquid between", *methane_liquid_k, 
      "K — Titan sits right in the rain zone")
```
