## 1. How far is far
Average distances from Earth, in millions of km (they change as planets orbit — these are rough means for the math).

```
import matplotlib.pyplot as plt

worlds = ["Moon", "Venus", "Mars", "Jupiter", "Saturn", "Neptune"]
dist_mkm = [0.384, 41, 78, 628, 1280, 4350]   # millions of km, rough averages
SPEED = 300000   # km per second — light, and radio

delay_min = [d * 1e6 / SPEED / 60 for d in dist_mkm]
for w, m in zip(worlds, delay_min):
    print(f"{w:8s}  one-way radio delay ≈ {m:8.1f} minutes")
```

```
plt.bar(worlds, delay_min, color="#7E1B14")
plt.ylabel("one-way delay (minutes)")
plt.title("Why nobody joysticks a Saturn probe")
plt.show()
```

## 2. The conversation test
A rover sees a cliff and asks Earth what to do. How long until the answer arrives? (Round trip = twice the delay.)

```
saturn = delay_min[4]
print("Saturn round trip:", round(2 * saturn), "minutes —",
      round(2 * saturn / 60, 1), "hours between question and answer")
```
