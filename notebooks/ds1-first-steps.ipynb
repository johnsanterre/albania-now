## 4. Text vs numbers — and the bridges
`7` is a number, `"7"` is text. `+` adds two numbers or glues two texts, and refuses to mix. `str()` and `int()` are the bridges.

```
print(7 + 7)        # math
print("7" + "7")    # gluing text
```

```
print("Age " + str(16))   # str() turns a number into text — now + can glue
```

```
mosha_text = "16"
print(int(mosha_text) + 1)   # int() goes the other way
```

```
print("Va" * 3)     # multiplying text repeats it — real, and really useful
```
