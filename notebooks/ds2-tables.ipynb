## 1. Load the table
pandas is Python's table tool. The table is 8 Albanian cities (populations rounded, for practice).

```
import pandas as pd
import io

csv = """city,region,population,coastal
Tirana,Central,560000,no
Durres,Coast,200000,yes
Vlore,Coast,130000,yes
Elbasan,Central,140000,no
Shkoder,North,135000,no
Fier,South,120000,no
Korce,Southeast,75000,no
Sarande,Coast,40000,yes
"""
df = pd.read_csv(io.StringIO(csv))
df
```

One row per city, one column per fact. `len(df)` counts rows — count before and after every filter.

```
len(df)
```

## 2. Filter — keep rows that pass a test

```
coastal = df[df["coastal"] == "yes"]
coastal
```

```
big = df[df["population"] > 100000]
print(len(df), "rows in,", len(big), "rows out")
big
```

## 3. Sort — same rows, new order

```
df.sort_values("population", ascending=False)
```

## 4. Group — fold the table and summarize each pile

```
df.groupby("region")["population"].mean().round(0)
```

```
df.groupby("region").size()
```
