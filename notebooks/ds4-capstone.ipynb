## 1. The data
One month of a small (fictional) Tirana shop: 120 rows — 30 days × 4 products. Generated, so every class gets the same table.

```
import pandas as pd

days = list(range(1, 31))
weekday = ["Mon","Tue","Wed","Thu","Fri","Sat","Sun"]
products = [("byrek", 120), ("coffee", 150), ("juice", 180), ("ice cream", 200)]
rows = []
for d in days:
    wd = weekday[(d - 1) % 7]
    for i, (prod, price) in enumerate(products):
        base = 20 + 8 * ((d * 7 + i * 13) % 5)
        if wd in ("Sat", "Sun"): base += 15
        if prod == "ice cream": base += d // 3   # warming month
        rows.append([d, wd, prod, base, price])
df = pd.DataFrame(rows, columns=["day", "weekday", "product", "units", "price_lek"])
df["revenue_lek"] = df["units"] * df["price_lek"]
print(len(df), "rows")
df.head()
```

## 2. The loop, walked once
**Question:** which weekday earns the most?

```
by_day = df.groupby("weekday")["revenue_lek"].mean().sort_values(ascending=False)
by_day
```

```
import matplotlib.pyplot as plt

by_day.plot(kind="bar", color="#7E1B14")
plt.ylabel("mean revenue (lek)")
plt.title("Mean daily revenue by weekday — one month, one shop")
plt.show()
```

**The finding, three sentences:**
1. *Measured:* mean daily revenue by weekday, one month, one shop, 120 rows.
2. *Found:* weekend days average the highest revenue.
3. *Not proven:* that this holds in other months, other shops, or that the weekend causes it — school holidays, weather, and tourism all moved together this month.
