## The build — the traced program
The longest program of the week. Trace it ON PAPER first — every line, the whiteboard after each store, the final output. Photograph the paper. THEN run it.

```
cmimet = [120, 150, 180, 90]
total = 0
me_i_shtrenjte = 0
for cmim in cmimet:
    total = total + cmim
    if cmim > me_i_shtrenjte:
        me_i_shtrenjte = cmim
mesatarja = total / len(cmimet)
print("total:", total)
print("max:", me_i_shtrenjte)
print("mesatarja:", mesatarja)
```
