## 1. Thirty days of starlight
Brightness measured every 72 minutes for a month (synthetic but realistic — same shape as Kepler data).

```
import numpy as np
import matplotlib.pyplot as plt

N = 600
t_days = np.arange(N) * 30 / N
rng = np.sin(np.arange(N) * 12.9898) * 43758.5453
noise = (rng - np.floor(rng) - 0.5) * 6          # the star's flicker
flux = 1000 + noise
for center in (80, 280, 480):                     # something passes in front…
    mask = np.abs(np.arange(N) - center) < 7
    flux[mask] -= 12

plt.figure(figsize=(10, 3.2))
plt.plot(t_days, flux, lw=0.8, color="#7E1B14")
plt.xlabel("time (days)")
plt.ylabel("brightness")
plt.title("One star, one month")
plt.show()
```

## 2. Detect the dips
A threshold, exactly like the figure on the lesson page:

```
THRESHOLD = 991

below = flux < THRESHOLD
starts = np.where(below & ~np.roll(below, 1))[0]
print("Dip starts at samples:", starts)
print("Dip starts at days:   ", np.round(t_days[starts], 1))
```

## 3. Period and depth

```
period_days = np.diff(t_days[starts]).mean()
depth = 1000 - flux[below].mean()
depth_frac = depth / 1000
print(f"Period ≈ {period_days:.1f} days — the planet's year")
print(f"Depth ≈ {depth:.1f} units = {depth_frac:.3%} of the light")
```

## 4. The size, from the square-root rule

```
radius_ratio = np.sqrt(depth_frac)
print(f"Planet radius ≈ {radius_ratio:.2%} of the star's radius")
print("For a Sun-like star that is roughly Jupiter-sized")
```
