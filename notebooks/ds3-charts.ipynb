## 1. The honest bar chart
Two shops: A sold 96, B sold 100.

```
import matplotlib.pyplot as plt

shops = ["Shop A", "Shop B"]
sales = [96, 100]

plt.bar(shops, sales, color=["#D8291F", "#7E1B14"])
plt.ylabel("items sold (July)")
plt.title("Honest: axis starts at 0")
plt.show()
```

## 2. The same numbers, lying
One line changes: where the axis starts.

```
plt.bar(shops, sales, color=["#D8291F", "#7E1B14"])
plt.ylabel("items sold (July)")
plt.ylim(90, 101)   # the chop
plt.title("Chopped: same data, different story")
plt.show()
```

## 3. A line chart — change over time

```
weeks = [1, 2, 3, 4]
visits = [120, 135, 128, 160]

plt.plot(weeks, visits, marker="o", color="#7E1B14")
plt.xlabel("week")
plt.ylabel("library visits")
plt.xticks(weeks)
plt.show()
```

## 4. A scatter — relationship
Each dot is one of last week's cities: population across, a (invented, for practice) count of high schools up.

```
pop = [560, 200, 130, 140, 135, 120, 75, 40]        # thousands
schools = [48, 19, 12, 14, 13, 11, 8, 4]

plt.scatter(pop, schools, color="#D8291F")
plt.xlabel("population (thousands)")
plt.ylabel("high schools")
plt.show()
```

The cloud leans — bigger cities, more schools. A relationship, not a cause.
