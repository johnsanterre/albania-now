## 1. Build a spectrum
Brightness across wavelength — smooth starlight, before any gas touches it.

```
import numpy as np
import matplotlib.pyplot as plt

wl = np.linspace(1.0, 5.0, 400)       # wavelength, micrometers (infrared)
star = 1.0 - 0.05 * (wl - 3.0) ** 2    # smooth starlight

plt.plot(wl, star, color="#7E1B14")
plt.xlabel("wavelength (micrometers)")
plt.ylabel("brightness")
plt.title("Starlight, untouched")
plt.show()
```

## 2. Pass it through methane
Methane steals light near 2.3 and 3.3 micrometers — its infrared fingerprint. We carve those dips in:

```
def dip(wl, center, width, depth):
    return depth * np.exp(-((wl - center) / width) ** 2)

seen = star - dip(wl, 2.3, 0.08, 0.25) - dip(wl, 3.3, 0.12, 0.45)

plt.plot(wl, star, color="#B9B2A6", label="before the gas")
plt.plot(wl, seen, color="#D8291F", label="after methane")
plt.xlabel("wavelength (micrometers)")
plt.ylabel("brightness")
plt.legend()
plt.title("The fingerprint")
plt.show()
```

## 3. Find the thief by code
Where is the deepest bite?

```
deepest = wl[np.argmin(seen)]
print("Deepest absorption at", round(deepest, 2), "micrometers")
```

That number — 3.3 — is how a spectrum names its gas: look up which molecule bites there, and methane is the answer. This is what Cassini's CIRS did at Titan, in the infrared, for thirteen years.
